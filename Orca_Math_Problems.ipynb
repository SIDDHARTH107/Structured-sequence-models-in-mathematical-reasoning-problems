```python
def extract_final_answer(text):
    if not isinstance(text, str):
        return ""

    # Trying to find numeric answers first
    matches = re.findall(r"[-+]?\d*\.\d+|\d+", text)
    if matches:
        return matches[-1]  # Returning last numeric match

    # If no number found, trying to get last word from the last sentence
    sentences = re.split(r"[.?!]", text)
    last_sentence = sentences[-2].strip() if len(sentences) > 1 else sentences[0].strip()
    return last_sentence.split(" ")[-1]
```

```python
df["final_answer"] = df["answer"].apply(extract_final_answer)
```
